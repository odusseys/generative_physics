# Airfoil Flux LoRA training

Condition images show the random foil in uniform rightward flow without streamlines. Target images show the potential-flow speed field and streamlines for the same foil.

```python
from pde_flux_lora.airfoil import generate_airfoil_image_pairs
import matplotlib.pyplot as plt
from IPython.display import display

pairs = generate_airfoil_image_pairs(range(5), sim_nx=256, output_size=256)

fig, axes = plt.subplots(len(pairs), 2, figsize=(6, 3 * len(pairs)))
for row, (flow_img, no_flow_img, params) in enumerate(pairs):
    axes[row, 0].imshow(no_flow_img)
    axes[row, 0].set_title(f"condition: no-flow seed={params['seed']}")
    axes[row, 1].imshow(flow_img)
    axes[row, 1].set_title("target: flow")
    for ax in axes[row]:
        ax.axis("off")
plt.tight_layout()
display(fig)
plt.close(fig)
```
